## 1. Import Library

```python
import pandas as pd
from dotenv import load_dotenv
import os
from sqlalchemy import create_engine
```

```python
# Untuk memunculkan semua kolom data
pd.set_option('display.max_columns', None)
```

## 2. Load Environment & Database Connection

```python
load_dotenv()

DB_HOST = os.getenv("DB_HOST")
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")
```

```python
engine = create_engine(f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}")
```

## 3. Load Dataset

```python
df = "SELECT * FROM superstore_db.superstore"

df = pd.read_sql(df, engine)
```

```python
# Convert kolom ke datetime 'order_date'
df['order_date'] = pd.to_datetime(df['order_date'])
df['order_date_only'] = df['order_date'].dt.date

# Rentang waktu data 'order_date'
print(f"Mulai   : {df['order_date_only'].min()}")
print(f"Selesai : {df['order_date_only'].max()}")
```

## Kesimpulan & Catatan Anomali EDA Profiling

Dari proses diagnosa data di atas, ditemukan beberapa poin penting sebelum masuk ke tahap *Data Cleansing*
1. **Tipe Data Tidak Sesuai:** Kolom numerik (`sales`, `quantity`, `discount`, `profit`) dan kolom waktu (`order_date`, `ship_date`) masih terbaca sebagai *string* karna bawaan dari *database*. Ini wajib di ubah ke tipe data yang benar.
2. **Kewajaran Angka Bisnis:** Ditemukan nilai `profit` yang minus cukup ekstrem hingga **-$6,599.98**. Setelah ditelusuri, ini logis dan wajar karena ada transaksi yang menggunakan `discount` yang sangat tinggi mencapai 80%.
3. **Karakter & Teks Kategorikal:** Data terbukti bersih dari *typo*, inkonsistensi huruf besar/kecil, ataupun spasi di ujung teks.
4. **Duplikasi Data:** Ditemukan 1 baris duplikat (identik di semua kolom selain `row_id`, yang dikecualikan karena nomor urut selalu unik). Baris ini akan dihapus pada tahap staging (1 dihapus, 1 dipertahankan).
5. **Kewajaran Waktu:** Rentang waktu transaksi berjalan secara logis dengan data pesanan *(Order Date)* dari **Januari 2014 sampai Desember 2017**, serta tanggal pengiriman *(Ship Date)* yang berjalan logis hingga **Januari 2018**. Jeda waktu pengiriman terpantau aman dan wajar.
6. **Format Tanggal Amerika:** Dataset ini menggunakan format Amerika, saat menggunakan SQL perlu disesuaikan.
